# 01. 데이터 로드·형식 정리

- 입력: `data/raw/SeoulBikeData.csv` (원본, 수정하지 않음)
- 하는 일: Latin-1 인코딩으로 읽기 → 컬럼명을 영문 소문자로 변경 → `date`를 datetime으로 변환
- 출력: `data/processed/seoul_bike_clean.csv` (UTF-8) → 02, 03에서 사용

```python
import sys
from pathlib import Path

# 프로젝트 루트(notebooks/의 상위 폴더)를 import 경로에 추가해서 src 패키지를 사용
ROOT = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.append(str(ROOT))

import pandas as pd

from src.config import RAW_PATH, CLEAN_PATH
```

```python
# 원본은 UTF-8이 아님
df = pd.read_csv(RAW_PATH, encoding='latin-1')
print(df.shape)
df.head()
```

```python
# 컬럼명을 영문 소문자로 변경 (°, 공백, 괄호, 단위 제거)
COLUMN_NAMES = {
    'Date': 'date',
    'Rented Bike Count': 'rented_bike_count',
    'Hour': 'hour',
    'Temperature(°C)': 'temperature',
    'Humidity(%)': 'humidity',
    'Wind speed (m/s)': 'wind_speed',
    'Visibility (10m)': 'visibility',
    'Dew point temperature(°C)': 'dew_point_temp',
    'Solar Radiation (MJ/m2)': 'solar_radiation',
    'Rainfall(mm)': 'rainfall',
    'Snowfall (cm)': 'snowfall',
    'Seasons': 'seasons',
    'Holiday': 'holiday',
    'Functioning Day': 'functioning_day',
}
df = df.rename(columns=COLUMN_NAMES)
df.columns.tolist()
```

```python
# Date는 dd/mm/yyyy 문자열 -> datetime
df['date'] = pd.to_datetime(df['date'], dayfirst=True)
print(df['date'].min(), '~', df['date'].max())
df.dtypes
```

```python
# 정리본을 UTF-8로 저장 후 다시 읽어서 확인
CLEAN_PATH.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(CLEAN_PATH, index=False, encoding='utf-8')

check = pd.read_csv(CLEAN_PATH, parse_dates=['date'])
print(check.shape)
print(check.equals(df))
```
